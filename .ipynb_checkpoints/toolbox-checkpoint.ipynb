```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import pandas as pd
import math 
import scipy.io as scio
import scipy as sc
import seaborn as sns
import cmocean as cmo
import cartopy.feature as cfeature
import cartopy.crs as ccrs
```

```python
def get_spatial_coverage(vr_grid):
    """
    get a map of spatial coverage of your data 
    shaped as (time,x,y)
    """    
    
    dim_t = vr_grid.shape[0]
    dim_x = vr_grid.shape[1]
    dim_y = vr_grid.shape[2]
    dim_s = dim_x*dim_y
    
    coverage = np.nan*np.ones((dim_t))
    vr_r = np.reshape(vr_grid, (dim_t,dim_s))
    for ts in range(dim_t):
        vr_ts = vr_r[ts]
        number_nonans = len(vr_ts[~np.isnan(vr_ts)])
        coverage[ts] = number_nonans/len(vr_ts)
    return coverage
```
