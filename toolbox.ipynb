# GRIDS, LON, LAT

```python
def get_spatial_coverage(vr_grid):
    """
    get a map of spatial coverage of your data 
    shaped as (time,x,y)
    """    
    
    dim_t = vr_grid.shape[0]
    dim_x = vr_grid.shape[1]
    dim_y = vr_grid.shape[2]
    dim_s = dim_x*dim_y
    
    coverage = np.nan*np.ones((dim_t))
    vr_r = np.reshape(vr_grid, (dim_t,dim_s))
    for ts in range(dim_t):
        vr_ts = vr_r[ts]
        number_nonans = len(vr_ts[~np.isnan(vr_ts)])
        coverage[ts] = number_nonans/len(vr_ts)
    return coverage
```

```python
def get_pixel_coverage(vr_grid):
    """
    get a map of temporal coverage of your data 
    shaped as (time,x,y)
    """
    
    dim_t = vr_grid.shape[0]
    dim_x = vr_grid.shape[1]
    dim_y = vr_grid.shape[2]
    
    map_coverage = np.nan*np.zeros((dim_x, dim_y))
    for i in range(dim_x):
        for j in range(dim_y):
            pixel = vr_grid[:,i,j]
            len_pix = len(pixel[~np.isnan(pixel)])
            map_coverage[i,j] = len_pix
            
    return map_coverage
```

# PLOTS

```python
import cartopy.mpl.ticker as cticker
def fig_cot_au(nrows,ncols):
    """
    creates a figure and adds the cartopy coastline and ocean 
    """
    
    fig, axs = plt.subplots(nrows=nrows,ncols=ncols,
                        subplot_kw={'projection': ccrs.PlateCarree()},
                        figsize=(11,8.5),sharex=True,sharey=True)

    # axs is a 2 dimensional array of `GeoAxes`.  We will flatten it into a 1-D array
    if nrows+ncols>2:
        axs=axs.flatten()

    for ax in fig.get_axes():
        gl = ax.gridlines(draw_labels=True,zorder=0)
        ax.add_feature(cfeature.OCEAN,zorder=1)
        ax.add_feature(cfeature.COASTLINE,zorder=2)
        
        gl.top_labels = False
        gl.right_labels = False
    return fig, axs
```
